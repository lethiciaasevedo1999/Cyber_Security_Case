```python
import boto3
import awswrangler as wr
import pandas as pd

boto3.setup_default_session(region_name="us-east-2")

DATABASE = "cyber_security_db"

def sql(query: str) -> pd.DataFrame:
    return wr.athena.read_sql_query(
        sql=query,
        database=DATABASE,
        ctas_approach=False,
    )
```
